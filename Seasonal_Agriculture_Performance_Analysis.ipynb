```python
season_data = df.groupby("Season").agg(
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Cost=("Total_Cost_INR", "mean"),
    Average_Profit=("Profit_INR", "mean"),
    Average_Water_Used=("Water_Used_m3", "mean")
)

season_data = season_data.round(2)

season_data
```

```python
# Find the best season based on average profit
best_season = season_data["Average_Profit"].idxmax()
best_profit = season_data.loc[best_season, "Average_Profit"]

print("Best season based on average profit:", best_season)
print("Average profit: ₹", round(best_profit, 2))
```

```python
# Visualise average profit for each season
plt.figure(figsize=(7, 4))

sns.barplot(
    data=season_data.reset_index(),
    x="Season",
    y="Average_Profit",
    color="teal"
)

plt.title("Average Profit by Season")
plt.xlabel("Season")
plt.ylabel("Average Profit (INR)")
plt.show()
```

The chart and table above show which season has the highest average profit. This result is calculated directly from the dataset.



# **Crop-wise Analysis**

---



This section compares crops using average yield, production, revenue, cost, and profit.

```python
crop_data = df.groupby("Crop").agg(
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Total_Production=("Production_Tonnes", "sum"),
    Average_Revenue=("Revenue_INR", "mean"),
    Average_Cost=("Total_Cost_INR", "mean"),
    Average_Profit=("Profit_INR", "mean")
)

crop_data = crop_data.round(2)
crop_data = crop_data.sort_values("Average_Profit", ascending=False)

crop_data
```

```python
# Find the crop with the highest average profit
best_crop = crop_data["Average_Profit"].idxmax()
best_crop_profit = crop_data.loc[best_crop, "Average_Profit"]

# Find the crop with the highest average yield
highest_yield_crop = crop_data["Average_Yield"].idxmax()
highest_yield = crop_data.loc[highest_yield_crop, "Average_Yield"]

print("Most profitable crop:", best_crop)
print("Average profit: ₹", round(best_crop_profit, 2))

print("\nCrop with highest yield:", highest_yield_crop)
print("Average yield:", round(highest_yield, 2), "tonnes/hectare")
```

```python
# Average profit by crop
plt.figure(figsize=(9, 5))

sns.barplot(
    data=crop_data.reset_index(),
    x="Average_Profit",
    y="Crop",
    color="mediumseagreen"
)

plt.title("Average Profit by Crop")
plt.xlabel("Average Profit (INR)")
plt.ylabel("Crop")
plt.show()
```

```python
# Average yield by crop
yield_data = crop_data.sort_values("Average_Yield", ascending=False)

plt.figure(figsize=(9, 5))

sns.barplot(
    data=yield_data.reset_index(),
    x="Average_Yield",
    y="Crop",
    color="orange"
)

plt.title("Average Yield by Crop")
plt.xlabel("Average Yield (Tonnes per Hectare)")
plt.ylabel("Crop")
plt.show()
```

### **Observation**

---



The crop with the highest yield is not always the crop with the highest profit.
Profit also depends on production cost and market price.

# **Resource and Environmental Analysis**

---



This section studies irrigation methods, water use, rainfall, temperature, and their relationship with crop yield.

```python
irrigation_data = df.groupby("Irrigation_Method").agg(
    Average_Water_Used=("Water_Used_m3", "mean"),
    Average_Water_Efficiency=("Water_Efficiency_t_per_1000m3", "mean"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Average_Profit=("Profit_INR", "mean")
)

irrigation_data = irrigation_data.round(2)
irrigation_data
```

```python
# Find the most water-efficient irrigation method
best_irrigation = irrigation_data["Average_Water_Efficiency"].idxmax()

print("Most water-efficient irrigation method:", best_irrigation)
print(
    "Water efficiency:",
    irrigation_data.loc[best_irrigation, "Average_Water_Efficiency"],
    "tonnes per 1,000 m³"
)
```

```python
# Compare water efficiency across irrigation methods
plt.figure(figsize=(8, 4))

sns.barplot(
    data=irrigation_data.reset_index(),
    x="Irrigation_Method",
    y="Average_Water_Efficiency",
    color="royalblue"
)

plt.title("Water Efficiency by Irrigation Method")
plt.xlabel("Irrigation Method")
plt.ylabel("Water Efficiency (Tonnes per 1,000 m³)")
plt.xticks(rotation=15)
plt.show()
```

# **Economic Analysis**

---



This section studies revenue, total cost, profit, and the relationship between important variables.

```python
# Overall financial summary
total_revenue = df["Revenue_INR"].sum()
total_cost = df["Total_Cost_INR"].sum()
total_profit = df["Profit_INR"].sum()

print("Total Revenue: ₹", round(total_revenue, 2))
print("Total Cost: ₹", round(total_cost, 2))
print("Total Profit: ₹", round(total_profit, 2))
```

```python
# Revenue versus cost graph
plt.figure(figsize=(7, 4))

sns.scatterplot(
    data=df,
    x="Total_Cost_INR",
    y="Revenue_INR",
    hue="Season",
    alpha=0.6
)

plt.title("Revenue Compared With Total Cost")
plt.xlabel("Total Cost (INR)")
plt.ylabel("Revenue (INR)")
plt.show()
```

## **Statistical Test**

---



The Kruskal-Wallis test checks whether profit differs significantly between seasons.

A p-value below 0.05 means the difference is statistically significant.

```python
from scipy.stats import kruskal

kharif_profit = df[df["Season"] == "Kharif"]["Profit_INR"]
rabi_profit = df[df["Season"] == "Rabi"]["Profit_INR"]
zaid_profit = df[df["Season"] == "Zaid"]["Profit_INR"]

test_statistic, p_value = kruskal(
    kharif_profit,
    rabi_profit,
    zaid_profit
)

print("Kruskal-Wallis test statistic:", round(test_statistic, 3))
print("P-value:", round(p_value, 5))

if p_value < 0.05:
    print("Result: Average profit differs significantly across seasons.")
else:
    print("Result: No significant profit difference was found across seasons.")
```
